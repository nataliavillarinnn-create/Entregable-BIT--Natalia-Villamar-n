```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
iris = sns.load_dataset("iris")
```

```python
iris
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
fig.suptitle("Análisis de tamaños por especie", fontsize=16)

#Gráfico 1
sns.boxplot(x="species", y="sepal_length", data=iris, hue='species', palette="Paired", ax=axes[0, 0])
axes[0, 0].set_title("Longitud del pétalo según la especie")

#Gráfico 2
sns.scatterplot(x="petal_length", y="petal_width", data=iris, hue='species', palette="Paired", ax=axes[0, 1])
axes[0, 1].set_title("Relación Largo vs Ancho del Pétalo", fontsize=12)
axes[0, 1].set_xlabel("Largo del pétalo (cm)")
axes[0, 1].set_ylabel("Ancho del pétalo (cm)")

#Gráfico 3
sns.histplot(x="petal_length", data=iris, hue='species', palette="Paired", ax=axes[1, 0])
axes[1, 0].set_title("Distribución de la Longitud del Pétalo", fontsize=12)
axes[1, 0].set_xlabel("Largo del pétalo (cm)")
axes[1, 0].set_ylabel("Frecuencia (Conteo)")

#Gráfico 4
sns.barplot(x="species", y="sepal_width", data=iris, hue="species", palette="Paired", ax=axes[1, 1])
axes[1, 1].set_title("Promedio del Ancho del Sépalo por Especie", fontsize=12)
axes[1, 1].set_xlabel("Especie")
axes[1, 1].set_ylabel("Ancho Promedio (cm)")


plt.tight_layout()
plt.show()
```
