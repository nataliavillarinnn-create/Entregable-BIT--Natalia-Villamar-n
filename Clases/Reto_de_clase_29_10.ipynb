```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
df = pd.DataFrame({
    "Mes": ["Ene", "Feb", "Mar", "Abr", "May", "Jun"],
    "Ventas": [120, 150, 135, 180, 210, 195],
    "Categoria": ["A", "B", "A", "C", "B", "C"],
    "Unidades": [20, 35, 25, 40, 30, 45]
})
#Gráfico de línea, ventas por mes.
plt.figure(figsize=(8, 5))
plt.plot(df['Mes'], df['Ventas'], marker='o', color='green', linestyle='dotted')
for i, txt in enumerate(df['Ventas']):
    plt.text(df['Mes'][i], df['Ventas'][i], txt, ha='center', va='bottom')
plt.title("Ventas por Mes")
plt.xlabel("Mes")
plt.ylabel("Ventas")
plt.grid()
plt.show()
```

```python
df = pd.DataFrame({
    "Mes": ["Ene", "Feb", "Mar", "Abr", "May", "Jun"],
    "Ventas": [120, 150, 135, 180, 210, 195],
    "Categoria": ["A", "B", "A", "C", "B", "C"],
    "Unidades": [20, 35, 25, 40, 30, 45]
})

# 3 desafio
sns.set_style("whitegrid")
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x='Categoria', y='Unidades', palette="deep")

plt.title("Unidades por Categoría")
plt.xlabel("Categoría")
plt.ylabel("Unidades")
plt.show()
```
