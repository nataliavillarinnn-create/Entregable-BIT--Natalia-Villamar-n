```python
df = df.drop_duplicates()
df.shape
```

```python
#Utiliza unique() o value_counts() para identificar categorías inconsistentes o valores sospechosos.
df['Animal type'].value_counts()
```

```python
#Correción, se encuentran duplicados lógicos, de manera que se realiza correción de escritura.
df['Animal type'] = df['Animal type'].replace({
    'red squirrell': 'red squirrel',
    'red squirel': 'red squirrel',

    'ledgehod': 'hedgehog',
    'wedgehod': 'hedgehog',

    'European bisson': 'European bison',
    'European buster': 'European bison',
    'European bison™': 'European bison',

    "lynx?": "lynx"
})

# Verificar
df['Animal type'].value_counts()
```

```python
df['Country'].value_counts()
```

```python
#Correción.
df["Country"] = df["Country"].replace({
    "PL" : "Poland",
    "HU" : "Hungry",
    "Hungry" : "Hungary",
    "CZ" : "Czech Republic",
    "CC" : "Czech Republic",
    "Czech" : "Czech Republic",
    "DE" : "Germany",
    "Australia" : "Austria"

    })

#Verificación
df['Country'].value_counts()
```

```python
df[(df['Longitude'] > 180) | (df['Longitude'] < -180)]
```

**Al ejecutar el comando se evidencia que no hay valores por fuera de los rangos establecidos, los cuales podrían ser errores.**

```python
df[(df['Weight kg'] < 0)]
```

```python
df[(df['Body Length cm'] < 0)]
```

```python
#Los valores negativos son remplazados por valores nulos.
df.loc[df['Weight kg'] < 0, 'Weight kg'] = np.nan
df.loc[df['Body Length cm'] < 0, 'Body Length cm'] = np.nan
```

```python
#Ahora se remplazaran los datos faltantes en peso, altura, latitud y lontitud por la mediana.
df['Weight kg'] = df['Weight kg'].fillna(df['Weight kg'].median())
df['Body Length cm'] = df['Body Length cm'].fillna(df['Body Length cm'].median())
df['Latitude'] = df['Latitude'].fillna(df['Latitude'].median())
df['Longitude'] = df['Longitude'].fillna(df['Longitude'].median())
```

```python
#Las celdas que se encuentren vacias, se remplazarán por "Desconocido"
columnas_texto = ['Animal type', 'Country', 'Gender', 'Animal name']
for col in columnas_texto:
    df[col] = df[col].fillna("Unknown")
```

```python
#Verificación de que ya no se encuentren celdas importantes, vacias. Al no tener una gran relevancia el código del animal, se decide no tratar.
df.isnull().sum()
```

```python
Bisontes_Europeos = df[df['Animal type'] == 'European bison']
Bisontes_Europeos = np.nanmean(Bisontes_Europeos['Weight kg'])
print(f"El promedio de peso de los bisontes europeos es: {Bisontes_Europeos:.2f} kg")

Ardilla_Roja = df[df['Animal type'] == 'red squirrel']
Ardilla_Roja = np.nanmean(Ardilla_Roja['Weight kg'])
print(f"El promedio de peso de las ardillas rojas es: {Ardilla_Roja:.2f} kg")

Erizo = df[df['Animal type'] == 'hedgehog']
Erizo = np.nanmean(Erizo['Weight kg'])
print(f"El promedio de peso de los erizos es: {Erizo:.2f} kg")

Lince = df[df['Animal type'] == 'lynx']
Lince = np.nanmean(Lince['Weight kg'])
print(f"El promedio de peso de los linces es: {Lince:.2f} kg")
```

```python
porcentaje_RS = np.mean(df['Animal type'] == 'red squirrel') * 100
porcentaje_H = np.mean(df['Animal type'] == 'hedgehog') * 100
porcentaje_L = np.mean(df['Animal type'] == 'lynx') * 100
porcentaje_EB = np.mean(df['Animal type'] == 'European bison') * 100


print(f"Porcentaje de ardilla roja observadas: {porcentaje_RS:.2f}%")
print(f"Porcentaje de erizos observados: {porcentaje_H:.2f}%")
print(f"Porcentaje de Linces observadss: {porcentaje_L:.2f}%")
print(f"Porcentaje de Bisontes Europeos observados: {porcentaje_EB:.2f}%")
```

### **Análisis con groupby()**

```python
#Cuál es el número de observaciones por tipo de animal?
Observados = df.groupby('Animal type').size()
print("Observaciones por tipo de animal")
print(Observados)
```

```python
# Cuál es el promedio de una variable numérica según el tipo de animal?
Altura_promedio = df.groupby('Animal type')['Body Length cm'].mean()
print("Peso promedio por tipo de animal")
print(Altura_promedio)
```

```python
#Cuál es el promedio de una variable numérica según el país?
Longitud= df.groupby('Country')['Body Length cm'].mean()
print("Longitud promedio según el país")
print(Longitud)
```

```python
Animales_por_pais = df.groupby('Country').size().sort_values(ascending=False)
print("Animales por país")
print(Animales_por_pais)
```
